```python
#ขั้นที่ 2 ออกแบบ
class DrinkOrder:
  def __init__(self, order_id, customer_name, menu_name, drink_type, size, topping, sweetness_level):
    self.order_id = order_id
    self.customer_name = customer_name
    self.menu_name = menu_name
    self.drink_type = drink_type
    self.size = size
    self.topping = topping
    self.sweetness_level = sweetness_level #ระดับความหวาน: 0%, 25%, 50%, 100%, 125% (ไม่คิดเงิน)
    self.status = "รอดำเนินการ"

  def calculate_price(self):
    base_prices = {
        # กาแฟ
        "อเมริกาโน่" : 45,
        "ลาเต้" : 50,
        "คาปูชิโน่" : 50,
        "เอสเพรสโซ่" : 50,
        "มอคค่า" : 50,

        # ชาและนม
        "ชาเขียว" : 45,
        "ชาไทย" : 45,
        "มัทฉะลาเต้" : 50,
        "ช็อกโกแลต" : 50,
        "นมสด" : 40,

        # โซดา
        "สตอว์เบอร์รี่โซดา" : 40,
        "บลูฮาวายโซดา" : 40,
        "ยูซุโซดา" : 40
    }

    type_prices = {
        "ร้อน" : 0,
        "เย็น" : 5,
        "ปั่น" : 10
    }

    size_prices = {
        "S" : 0,
        "M" : 10,
        "L" : 20
    }

    topping_prices = {
        "ไม่มี" : 0,
        "ไข่มุก" : 5,
        "วิปครีม" : 15,
        "ครีมชีส" : 20
    }

    total = base_prices[self.menu_name] + type_prices[self.drink_type] + size_prices[self.size] + topping_prices[self.topping]
    return total

  def make_done(self):
   self.status = "เสร็จแล้ว"
```

```python
#ขั้นที่ 3 ฟังก์ชันสุ่มข้อมูล
import random

def generation_customer_name():
  """สุ่มชื่อลูกค้าจำลอง โดยรวมชื่อ+นามสกุลแบบสุ่ม"""
  first_names =["เจสซี่","ลีโอ","มายา","อเล็กซ์","คลาร่า","ลูคัส","นิโคล"]
  last_names =["รักเรียน","ใจดี","สายทอง","รุ่งเรือง","มั่นคง","วงศ์สว่าง","เจริญพร"]
  return random.choice(first_names) + " " + random.choice(last_names)

def random_price(min_price=40.0, max_price=50.0):
  """สุ่มราคากาแฟ/เครื่องดื่ม ระหว่าง min_price ถึง max_price บาท (ปัดทศนิยม 2 ตำแหน่ง)"""
  return round(random.uniform(min_price, max_price),2)

def random_sweetness_level():
  """สุ่มระดับความหวานของเครื่องดื่ม"""
  levels = ["0%","25%","50%","75%","100%"]
  return random.choice(levels)
```

```python
# ขั้นที่ 4 Loop
random.seed(1)

orders = []

menu_names = ["อเมริกาโน่", "ลาเต้", "คาปูชิโน่", "มอคค่า", "ชาเขียว"]
drink_types = ["กาแฟ", "ชา"]
sizes = ["S", "M", "L"]
toppings = ["ไม่มี", "ไข่มุก", "วิปครีม", "เจลลี่"]

for i in range(1, 301):

    customer_name = generation_customer_name()
    menu_name = random.choice(menu_names)
    drink_type = random.choice(drink_types)
    size = random.choice(sizes)
    topping = random.choice(toppings)
    sweetness_level = random_sweetness_level()

    order = DrinkOrder(
        i,
        customer_name,
        menu_name,
        drink_type,
        size,
        topping,
        sweetness_level
    )

    orders.append(order)

print(f"จำลองออเดอร์เสร็จแล้วทั้งหมด {len(orders)} รายการ")
```
